```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 100

df = pd.DataFrame({
    "fecha": pd.date_range(start="2023-01-01", periods=n, freq="D"),
    "ventas": np.random.randint(100, 1000, n)
})
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
plt.figure()

plt.plot(df["fecha"], df["ventas"])

plt.title("Evolución de ventas en el tiempo")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.grid()
linestyle="--"

plt.xticks(rotation=45)

plt.show()
```

# Insight:
# Se observa que las ventas fluctúan a lo largo del tiempo.
# Existen picos en ciertos días que podrían analizarse más a fondo.
# Este comportamiento puede indicar patrones de demanda.

```python
df["media_movil"] = df["ventas"].rolling(window=7).mean()

plt.figure()

plt.plot(df["fecha"], df["ventas"], label="Ventas")
plt.plot(df["fecha"], df["media_movil"], label="Media móvil", linestyle="--")
plt.grid()



plt.legend()

plt.title("Ventas y tendencia")
plt.show()
```
