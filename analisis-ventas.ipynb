## 📊 Análisis de ventas en el tiempo
Se agrupan las ventas por fecha para visualizar tendencias.

```python
import matplotlib.pyplot as plt

ventas_por_fecha = df2.groupby('Date')['Quantity Sold (kilo)'].sum()

ventas_por_fecha.plot(figsize=(10,5))
plt.title("Ventas a lo largo del tiempo")
plt.xlabel("Fecha")
plt.ylabel("Cantidad vendida")
plt.grid()

plt.show()
```

### 📊 Interpretación

Se observa cómo varían las ventas a lo largo del tiempo.
Existen ciertos picos que podrían estar relacionados con alta demanda o eventos específicos.

Esto sugiere que el comportamiento de ventas no es constante y puede depender de factores externos.

## 📊 Análisis de ingresos

En esta sección se calculan los ingresos totales a partir de la cantidad vendida y el precio unitario, y se analiza su evolución en el tiempo.

```python
df2["ventas"] = df2["Quantity Sold (kilo)"] * df2["Unit Selling Price (RMB/kg)"]
# calcular ingresos totales por fecha
ingresos_por_fecha = df2.groupby('Date')['ventas'].sum()

ingresos_por_fecha.plot(figsize=(10,5))
plt.title("Ingresos a lo largo del tiempo")
plt.xlabel("Fecha")
plt.ylabel("Ingresos")
plt.grid()

plt.show()
```

### 💰 Interpretación de ingresos

Los ingresos muestran el comportamiento económico del negocio.
Se pueden comparar con la cantidad vendida para entender si los picos se deben a mayor volumen o a precios más altos.

## 📈 Insights

- Las ventas presentan variaciones a lo largo del tiempo
- Se detectan picos que podrían indicar alta demanda
- Existen valores faltantes que deben tratarse antes de modelos predictivos
- Se creó una nueva variable de ingresos (ventas)
