```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

n = 100

df = pd.DataFrame({
    "edad": np.random.randint(18, 60, n),
    "ingreso_total": np.random.randint(100, 1000, n),
    "genero": np.random.choice(["Masculino", "Femenino"], n)
})
```

```python
df.head()
```

```python
df["grupo_edad"] = pd.cut(df["edad"], bins=[18,30,40,50,60])
```

```python
df.groupby("grupo_edad")["ingreso_total"].mean()
```

```python
df.groupby("genero")["ingreso_total"].mean()
```

```python
df.groupby("grupo_edad")["ingreso_total"].mean().plot(kind="bar")
plt.title("Ingreso promedio por edad")
plt.legend()
plt.show()
```

```python
df.groupby("genero")["ingreso_total"].mean().plot(kind="bar")
plt.title("Ingreso promedio por género")
plt.legend()
plt.show()
```

## 📈 Insights

- El grupo de edad que más ingresos genera es X
- Se observa diferencia de gasto entre géneros
- Los clientes jóvenes gastan menos en promedio
